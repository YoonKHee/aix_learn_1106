```python
# 데이터 전처리
fish = pd.read_csv('data/fish_logistic.csv')
fish.head(2)

data = fish.drop(columns=['Species']).values
label = fish['Species'].values
data,label

# 데이터 분리
train_data,test_data,train_label,test_label = train_test_split(data,label,random_state=42)
```

```python
# 데이터 정규화
ss = StandardScaler()
ss.fit(train_data)
train_scaled = ss.transform(train_data)
test_scaled = ss.transform(test_data)
```

```python
# 데이터 학습
lr = LogisticRegression(C=20,max_iter=1000)
lr.fit(train_scaled,train_label)

# # 소프트 맥스
# decisions = lr.decision_function(test_scaled[:5])
# proba = softmax(decisions,axis=1)
# print(lr.classes_)
# print(np.round(proba, decimals=3)) 

# 데이터 예측
predict = lr.predict(test_scaled)
predict1 = lr.predict_proba(test_scaled)
print(lr.classes_)
print(np.round(predict1[:3], decimals=3))

# 정확도
score = lr.score(train_scaled,train_label)
score1 = lr.score(test_scaled,test_label)

score,score1
```

```python
# 예측
in_data = '242,25.4,30,11.52,4.02'
in_data = list(map(float,in_data.split(',')))
in_data = ss.transform([in_data])

print(lr.classes_)
result = lr.predict_proba(in_data)
result = result.max()

r_result = lr.predict(in_data)
result,r_result

print(f'{result*100:.2f}확률로,{r_result}입니다')
```
