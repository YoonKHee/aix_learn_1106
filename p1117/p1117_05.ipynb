```python
import joblib
import os
from pathlib import Path
import numpy as np
from scipy.special import softmax

model_ex = joblib.load('model_ex.joblib')
```

```python
in_data=input('길이,높이,넓이를 입력하세요. 예)290.0,26.3,31.2,12.48,4.3056')
in_data = list(map(float,in_data.split(',')))
print(in_data)
```

```python
result = model_ex.predict([in_data])
print('예측생선',result)
result2 = model_ex.predict_proba([in_data])
result3 = result2.max()*100
print(f'예측확률:{result3:.2f}%')
print('전체확률:',(np.round(result2,decimals=3))*100)
model_ex.classes_
```
