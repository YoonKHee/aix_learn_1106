```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False


import pandas as pd
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor 
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.preprocessing import PolynomialFeatures,StandardScaler,OneHotEncoder 
from sklearn.metrics import mean_absolute_error # 평균절대값오차 오차범위
from sklearn.model_selection import train_test_split
import numpy as np
from io import StringIO
```

```python
df = pd.read_csv('data/manhattan.csv',)

df2 = df.columns[1:-2]
df
```

```python
data = df[df.columns[2:-2]].values
label = df['rent'].values

```

```python
train_data,test_data,train_label,test_label = train_test_split(data,label,random_state=42)
train_data[:2]
```

```python
# 데이터 학습
lr = LinearRegression()
lr.fit(train_data,train_label)

# 데이터 예측
predict = lr.predict(test_data)

# 정확도
score = lr.score(train_data,train_label)
score1 = lr.score(test_data,test_label)
score,score1
```

```python
poly = PolynomialFeatures(degree=3,include_bias=False)
poly.fit(train_data)

train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)

# 데이터 학습
lr = LinearRegression()
lr.fit(train_poly,train_label)

# 데이터 예측
predict = lr.predict(test_poly)
scor = lr.score(train_poly,train_label)
scor1 = lr.score(test_poly,test_label)
scor,scor1
```
