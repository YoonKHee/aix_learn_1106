```python
from sklearn import model_selection,svm,metrics
import pandas as pd
import numpy as np
from PIL import Image,ImageOps
```

```python
# 데이터 가져오기
train_csv = pd.read_csv('data/train.csv',header=None)
test_csv = pd.read_csv('data/t10k.csv',header=None)
```

```python
# 픽셀의 정규화 / 0-255 숫자를 0-1사이에 숫자로 정규화 
def test(x):
    output = []
    for i in x:
        output.append(float(i)/255)
    return output
```

```python
train_csv.iloc[0]
```

```python
# 데이터 0-255 /   0 0 0 0 255 253 7 0 0 0 0
train_csv_data = list(map(test,train_csv.iloc[:,1:].values))
test_csv_data = list(map(test,test_csv.iloc[:,1:].values))
```

```python
# 정답 label 작업
train_csv_label = train_csv[0].values
train_csv_label
test_csv_label = test_csv[0].values
test_csv_label
```

```python
# 데이터 학습
clf = svm.SVC() # 다른 알고리즘을 사용해도됨.
clf.fit(train_csv_data,train_csv_label)

# 데이터 예측
result = clf.predict(test_csv_data)
# 정답률
score =clf.score(test_csv_data,test_csv_label)
# score = metrics.accuracy_score(test_csv_label,result)
print('정답률',score)
```

```python
# 이미지 열기, 흑백으로 변환
img = Image.open('data/seven.png').convert("L")
# 색상반전
img = ImageOps.invert(img)
# ※ 학습 데이터는 검은 배경에 흰 글씨라 색을 뒤집어 맞춤
```

```python
# 숫자가 있는 영역만 추출
bbox = img.getbbox()
img = img.crop(bbox)
# 비율 축소 20*20
img.thumbnail((20,20),Image.Resampling.LANCZOS)
# 배경사이즈 28*28
new_img = Image.new('L',(28,28),0)
# 배경사이즈 숫자를 가운데 붙여넣기 (합친다)
x = (28-img.width)//2
y = (28-img.height)//2
new_img.paste(img,(x,y))
```

```python
# 예측용 데이터 - 0-255 / 0-1
img_data = np.array(new_img,dtype=np.float32)
float_img_data = img_data.reshape(1,784)/255.0
# ※ 28×28 그림을 한 줄(784칸)로 펴기
# float_img_data
```

```python
# 데이터 예측 - 2차원 형태여야 예측이 됨.
result = clf.predict(float_img_data)
print('확인:',result)
```
