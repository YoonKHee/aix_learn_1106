```python
import json
import pandas as pd
import requests
```

```python
df = pd.read_json('data/searchMovieList.json')
df
```

```python
movies = df.loc['movieList', 'movieListResult']
movies
```

```python
movie_df = pd.DataFrame(movies)
movie_df
directors = []
companys = []

for d in movie_df['directors']:
    if len(d) > 0:
        directors.append(d[0]['peopleNm'])
    else:
        directors.append('')

for c in movie_df['companys']:
    if len(c) > 0:
        companys.append(c[0]['companyNm'])
    else:
        companys.append('')

movie_df['directors'] = directors
movie_df['companys'] = companys

movie_df.to_csv('data/movie_list.csv', index=False, encoding='utf-8-sig')
```
