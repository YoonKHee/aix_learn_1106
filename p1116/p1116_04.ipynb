```python
df= pd.read_csv('data/iris.csv',index_col='caseno')
df

```

```python
# 데이터 전처리
data = df.drop(columns=['Species']).values
label = df['Species']
data,label
train_data,test_data,train_label,test_label = train_test_split(data,label,test_size=0.2,random_state=42)
```

```python
rf = RandomForestClassifier(n_estimators=100)
rf.fit(train_data,train_label)

print('train_정확도',rf.score(train_data,train_label))
print('test_정확도',rf.score(test_data,test_label))
```

```python
rf = RandomForestClassifier(n_jobs=-1,random_state=42)


scores = cross_validate(rf,train_data,train_label,return_train_score=True,n_jobs=-1)

print(scores['train_score'])
print(scores['test_score'])
print(np.mean(scores['train_score']))
print(np.mean(scores['test_score']))
print(rf.feature_importances_)
```
