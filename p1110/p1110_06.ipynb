```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False
import pandas as pd
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor 
from sklearn.metrics import mean_absolute_error # 평균절대값오차 오차범위
from sklearn.preprocessing import StandardScaler# 정규화 클래스
from sklearn.model_selection import train_test_split
import numpy as np
from io import StringIO
```

```python
# Quiz
# 결혼식장에 참석인원이 식사를하고가는 사람을 파악해서 음식을 준비하려고 합니다.
# 1. 참석인원(total), 식사인원(reception)을 학습하여 참석 300명일때 식사인원을 예측
# 2. 산점도그래프와 직선그래프를 출력하시오
```

```python
df = pd.read_csv('data/QuizData.csv')
```

```python
data = df.iloc[:,0].values
data.reshape(-1,1)[:2]
```

```python
data = df.iloc[:,0].values
label = df.iloc[:,1].values

train_data,test_data,train_label,test_label = train_test_split(data,label)
train_data = train_data.reshape(-1,1)
test_data = test_data.reshape(-1,1)
train_data
```

```python
# 학습
lr = LinearRegression()
lr.fit(train_data,train_label)

# 예측
predict = lr.predict(test_data)
predict

# 정답률
score = lr.score(train_data,train_label)
score2 = lr.score(test_data,test_label)
score,score2
```

```python
# 산점도 그래프
p = input('참석인원:')
p = float(p)
s = lr.predict([[p]])
print('식사인원:',s)

plt.scatter(train_data,train_label)
plt.scatter(p,s,marker='^')
plt.plot([10,500],[10*lr.coef_+lr.intercept_,500*lr.coef_+lr.intercept_])
plt.xlabel('참석인원')
plt.ylabel('식사인원')
plt.show()
```
