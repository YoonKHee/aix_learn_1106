```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
# matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus']=False


import pandas as pd
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor 
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression,SGDClassifier
from sklearn.preprocessing import PolynomialFeatures,StandardScaler,OneHotEncoder
from sklearn.metrics import mean_absolute_error # 평균절대값오차 오차범위
from sklearn.model_selection import train_test_split,cross_validate
from sklearn.tree import DecisionTreeClassifier,plot_tree
from scipy.special import expit, softmax # expit:이진분류함수, softmax:다중분류 함수 0-1숫자로 변환
import numpy as np
from io import StringIO
```

```python

# 데이터 가져오기, 전처리
wine = pd.read_csv('data/wine.csv')
wine.head(2)
data = wine[wine.columns[:-1]].values
label = wine['class'].values
data

# 데이터 분리
train_data,test_data,train_label,test_label = train_test_split(data,label,test_size=0.2,random_state=42)
sub_data,sub2_data,sub_label,sub2_label = train_test_split(train_data,train_label,test_size=0.25,random_state=42)
```

```python

dt = DecisionTreeClassifier(random_state=42,max_depth=3)
# 훈련데이터를 5등분(cv=5) -? 5번실행 # 디폴트 5
# 훈련데이터를 3등분(cv=3) -? 3번실행
# default:test score만 출력, return_train_score = True : train_sclre,test score 모두 출력
score = cross_validate(dt,train_data,train_label,cv=10,return_train_score=True)
# fit_time : 훈련시간, score-item:score생성시간, test_score: test정확도, trian_score : train정확도
score
```

```python
print('train_평균정확도',np.mean(score['train_score']))
print('test_평균정확도',np.mean(score['test_score']))
```
